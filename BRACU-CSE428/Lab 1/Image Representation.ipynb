```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, img_as_float
```

# **Gray Scale Image**

```python
img_grayscale = io.imread("/content/flower.jpg")
io.imshow(img_grayscale)
```

```python
print(img_grayscale)
```

```python
print("Height and Width of the image is: ", img_grayscale.shape)
```

```python
print("Maximum Pixel Value: ", img_grayscale.max())
print("Minimum Pixel Value: ",img_grayscale.min())

n_levels = len(np.unique(img_grayscale))
print("Number of pixel levels present:", n_levels)
```

```python
img_grayscale = img_grayscale / 255 #Normalizing the image

print("Maximum Pixel Value: ", img_grayscale.max())
print("Minimum Pixel Value: ",img_grayscale.min())

n_levels = len(np.unique(img_grayscale))
print("Number of pixel levels present:", n_levels)
```

# **RGB Image**

```python
img_rgb = io.imread("/content/sample_1.jpg")
io.imshow(img_rgb)
```

```python
print(img_rgb)
```

```python
height, width, nchannels = img_rgb.shape
print("Height =", height)
print("Width =", width)
print("Number of channels =", nchannels)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Red channel
axes[0].imshow(img_rgb[:, :, 0], cmap="grey")
axes[0].axis("off")
axes[0].set_title("Red Channel")

# Green channel
axes[1].imshow(img_rgb[:, :, 1], cmap="grey")
axes[1].axis("off")
axes[1].set_title("Green Channel")

# Blue channel
axes[2].imshow(img_rgb[:, :, 2], cmap="grey")
axes[2].axis("off")
axes[2].set_title("Blue Channel");
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Red channel
axes[0].imshow(img_rgb[:, :, 0], cmap="Reds")
axes[0].axis("off")
axes[0].set_title("Red Channel")

# Green channel
axes[1].imshow(img_rgb[:, :, 1], cmap="Greens")
axes[1].axis("off")
axes[1].set_title("Green Channel")

# Blue channel
axes[2].imshow(img_rgb[:, :, 2], cmap="Blues")
axes[2].axis("off")
axes[2].set_title("Blue Channel");
```
