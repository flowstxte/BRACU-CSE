```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/bank-additional-full.csv', sep=';')
print("Loaded:", df.shape)
```

```python
df = df.drop(columns=['duration'])
print("duration dropped —", df.shape)
```

```python
df['was_previously_contacted'] = (df['pdays'] != 999).astype(int)
df['pdays_clean'] = df['pdays'].replace(999, np.nan)
df['pdays_clean'] = df['pdays_clean'].fillna(0)

print(df[['pdays', 'was_previously_contacted', 'pdays_clean']].describe(include='all'))
df['was_previously_contacted'].value_counts()
```

```python
df.to_csv('../data/processed/bank_cleaned.csv', index=False)
print("Saved cleaned data:", df.shape)
```
